```python
#file handling
f=open("python.txt","w")
f.write("Hello World")
print("file created")
f.close()
```

```python
f=open("python.txt","r")
print(f.read())
f.close()
```

```python
with open("python.txt","a") as f:
  f.writelines(["\nProgram","\nExecuted"])
print("file updated")
```

```python
with open("python.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(0)
  print(f.tell())
  print(f.readlines())
  f.seek(0)
  print(f.read())
```

```python
#write a program that copy the content from one file to another file
f=open("python.txt","r")
data=f.read()
f.close()
file=open("new file.txt","w")
file.write(data)
print("content copied to new file")
file.close()
```

```python
#write a python program that prints each line of a file in reverse order
with open("python.txt","r") as f:
  for line in f:
    print(line[::-1])
```

```python
#program that converts all the characters into lower case and write in the order
with open("python.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  data_sorted=sorted(data_lower)
with open("new.txt","w") as f1:
  for i in data_sorted:
    f1.write(i+"\n")
with open("new.txt","r") as f2:
  print(f2.read())
```

```python
import pickle
with open("pickel.bin", "wb") as f:
    data = [1, 2, 3, "hello"]
    pickle.dump(data, f)
with open("pickel.bin", "rb") as f1:
    data_read = pickle.load(f1)
print(data_read)
```

```python
with open("images.jpg","rb") as f:
  data=f.read()
with open("n.jpg","wb") as f1:
  f1.write(data)
with open("n.jpg","rb") as f2:
  print(f2.read())
```
