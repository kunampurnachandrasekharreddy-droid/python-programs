```python
#list operations
#concatenation
L1=[1,2,3,"hello"]
L2=[11,22,33,"hai"]
print(L1+L2)
#repetition
print(L1*2)
#indexing
print(L1[3])
print(L1[0])
#slicing
print(L1[::2])
print(L1[1:3])
print(L1[::-1])
#membership operator
if 11 in L2:
  print("yes")
else:
  print("no")
if 43 in L2:
  print("yes")
else:
  print("no")
#iteration
for i in L2:
  print(i)
for i in range(len(L1)):
  print(L1[i])
```

```python
#Built in functions for list
L3=[22,33,44,55,11]
print(len(L3))
print(sum(L3))
print(min(L3))
print(max(L3))
print(list(reversed(L3)))
print(sorted(L3))
```

```python
#list methods
L4=[1,2,3,"hey"]
#append
L4.append("purna")
print(L4)
#extend
L4.extend(L2)
print(L4)
#insert
L4.insert(0,"python")
print(L4)
#pop
L4.pop()
print(L4)
L4.pop(2)
print(L4)
#index
print(L4.index("hey"))
#count
print(L4.count("hey"))
#remove
L4.remove("hey")
print(L4)
#reverse
L4.reverse()
print(L4)
```

```python
#program that prints the sum of list elements without using sum()
L=[1,9,8,4,6]
sum=0
for i in L:
  sum+=i
print(sum)
```
